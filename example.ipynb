```python
# 生成一个aritificial data跑一下

# quadlogit.demo.GenData可以简化



#生成数据
G,Xmat,Zmat,density,degree,transitivity,separation,trueParameter=quadlogit.demo.GenData(N=100, directed=True, mutual=False,specification="A1",seed=111)
# Xmat = Xmat.repeat(5, axis = -1)    # shape (N, N, K), K=5

#计算
quadlogit.fit(G,Xmat)
```

```python
#计算花时间的是loadindex，不管什么数据，给定100个以内网络nodes/agens下，可以调用包中函数结果预计算
import numpy as np
indices = quadlogit.generate_quad_indices(np.shape(G)[0])
```

```python
# 后续计算速度快很多。

res3 = quadlogit.fit(G,Xmat, indices=indices)
```
